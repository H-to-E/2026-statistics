```python
from sklearn.datasets import fetch_openml
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

X,y= fetch_openml("adult",version=2,as_frame=True,return_X_y=True)

df=X.copy()
df['income']=y
df=df[["age","workclass","education","education-num","marital-status","occupation","race",'sex','capital-gain','capital-loss','hours-per-week','income']]
df=df.dropna()
df
```

```python
print('\n PMF of occupation')
occupation_dist=df['occupation'].value_counts(normalize=True)
print(occupation_dist)

plt.figure()
occupation_dist.plot(kind='bar')
plt.title('PMF of occupation')
plt.xlabel('occupation')
plt.ylabel('Probability')
plt.xticks(rotation=90)
plt.show()
```

```python
print('\n PMF of race')
race_dist=df['race'].value_counts(normalize=True)
print(race_dist)

plt.figure()
race_dist.plot(kind='bar')
plt.title('PMF of race')
plt.xlabel('race')
plt.ylabel('Probability')
plt.xticks(rotation=90)
plt.show()
```

```python
sns.kdeplot(df["education-num"], color='red', fill=True)
plt.title('PDF of Education Number')
plt.xlabel('Education Number')
plt.ylabel('Density')
plt.show()
```

```python
sns.kdeplot(df["capital-loss"], color='red', fill=True)
plt.title('PDF of Capital Loss')
plt.xlabel('Capital Loss')
plt.ylabel('Density')
plt.show()
```
