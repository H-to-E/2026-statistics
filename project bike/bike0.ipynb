```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
seoul_bike_sharing_demand = fetch_ucirepo(id=560) 
  
# data (as pandas dataframes) 
X = seoul_bike_sharing_demand.data.features 
y = seoul_bike_sharing_demand.data.targets 
  
# metadata 
print(seoul_bike_sharing_demand.metadata) 
  
# variable information 
print(seoul_bike_sharing_demand.variables) 

df = pd.concat([X, y], axis=1)
df.head()

```

```python
#EDA

sns.histplot(x="Rented Bike Count", data=df)
plt.xlabel("Rented Bike Count")
plt.ylabel("Frequency")
plt.title("Histogram of Rented Bike Count")
plt.show()

sns.histplot(x="Hour", data=df)
plt.xlabel("Hour")
plt.ylabel("Frequency")
plt.title("Histogram of Hour")
plt.show()

sns.histplot(x="Temperature", data=df)
plt.xlabel("Temperature")
plt.title("Histogram of Temperature")
plt.show()


sns.histplot(x="Humidity", data=df)
plt.xlabel("Humidity")
plt.ylabel("Frequency")
plt.title("Histogram of Humidity")
plt.show()

sns.histplot(x="Wind speed", data=df)
plt.xlabel("Wind speed")
plt.title("Histogram of Wind Speed")
plt.show()


sns.histplot(x="Visibility", data=df)
plt.xlabel("Visibility")
plt.ylabel("Frequency")
plt.title("Box Plot of Visibility")
plt.show()

sns.histplot(x="Snowfall", data=df)
plt.xlabel("Snowfall")
plt.title("Histogram of Snowfall")
plt.show()

sns.histplot(x="Rainfall", data=df)
plt.xlabel("Rainfall")
plt.title("Histogram of Rainfall")
plt.show()

sns.histplot(x="Seasons", data=df)
plt.xlabel("Seasons")
plt.title("Histogram of Seasons")
plt.show()

sns.histplot(x="Holiday", data=df)
plt.xlabel("Holiday")
plt.title("Histogram of Holiday")
plt.show()

sns.histplot(x="Functioning Day", data=df)
plt.xlabel("Functioning Day")
plt.title("Histogram of Functioning Day")
plt.show()


```

```python
#사건

#자전거를 빌린 계절이 봄인 경우
s_rebtal=df[df["Seasons"]=="Spring"]["Rented Bike Count"].sum()/df["Rented Bike Count"].sum()
print("자전거를 빌린 계절이 봄인 경우:", s_rebtal.round(4))

#자전거를 2000대 이상 빌린 경우
s_rebtal=df[df["Rented Bike Count"]>=2000]["Rented Bike Count"].sum()/df["Rented Bike Count"].sum()
print("자전거를 2000대 이상 빌린 경우:", s_rebtal.round(4))

#기온이 20도 이상인 경우
s_rebtal=df[df["Temperature"]>=20]["Rented Bike Count"].sum()/df["Rented Bike Count"].sum()
print("기온이 20도 이상인 경우:", s_rebtal.round(4))    
```
