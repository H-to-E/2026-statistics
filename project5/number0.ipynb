```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint,bernoulli,binom

np.random.seed(42)

```

```python
low,high=1,7
n_samples=1000

uniform_samples = np.random.randint(low,high,size=n_samples)
sample_mean = np.mean(uniform_samples)
sample_var = np.var(uniform_samples)

a,b=low,high-1
theoretical_mean = (a+b)/2
theoretical_var = ((b-a+1)**2 - 1)/12

print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')
```

```python
x_uniform=np.arange(low,high)
pmf_uniform=randint.pmf(x_uniform,low,high)

plt.figure(figsize=(8,4))
plt.hist(uniform_samples,bins=np.arange(low-0.5,high+0.5,1),density=True,alpha=0.6,edgecolor='black')
plt.plot(x_uniform,pmf_uniform,'ro',label='Theoretical PMF')
plt.xticks(x_uniform)
plt.title('Discrete Uniform Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()
```
