```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint,bernoulli,binom

np.random.seed(42)

```

```python
n,p=10,0.4
n_samples=1000

binomial_samples = np.random.binomial(n, p, size=n_samples)
sample_mean = np.mean(binomial_samples)
sample_var = np.var(binomial_samples)

theoretical_mean = n * p
theoretical_var = n * p * (1 - p)

print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')
```

```python
x_binom=np.arange(0,n+1)
pmf_binom=binom.pmf(x_binom,n,p)

plt.figure(figsize=(8,4))
plt.hist(binomial_samples,bins=np.arange(-0.5,n+1.5,1),density=True,alpha=0.6,edgecolor='black')
plt.plot(x_binom,pmf_binom,'ro',label='Theoretical PMF')
plt.xticks(x_binom)
plt.title('Discrete Binomial Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()
```

```python
print(f'P(X=3)={binom.pmf(3,n,p):.4f}')
print(f'P(X<=3)={binom.cdf(3,n,p):.4f}')
print(f'P(X>=7)={binom.sf(6,n,p):.4f}')
```
