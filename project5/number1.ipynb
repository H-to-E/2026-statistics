```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint,bernoulli,binom

np.random.seed(42)

```

```python
p=0.3
n_samples=1000

bernoulli_samples=np.random.binomial(1,p,size=n_samples)



sample_mean = np.mean(bernoulli_samples)
sample_var = np.var(bernoulli_samples)

theoretical_mean = p
theoretical_var = p * (1 - p)


print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')
```

```python
x_bern=np.array([0,1])
pmf_bern=bernoulli.pmf(x_bern,p)

plt.figure(figsize=(8,4))
values, counts = np.unique(bernoulli_samples, return_counts=True)
plt.bar(values, counts/n_samples,width=0.5,alpha=0.6,edgecolor='black',label='Sample')
plt.plot(x_bern,pmf_bern,'ro',label='Theoretical PMF')
plt.xticks(x_bern)
plt.title('Discrete Bernoulli Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()
```
