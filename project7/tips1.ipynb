```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips=sns.load_dataset("tips")

sns.histplot(x="total_bill", y="tip", data=tips)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Histogram of Total Bill vs Tip")
plt.show()
```

```python
tips_df=tips[['total_bill', 'tip']]
tips_corr=tips_df.corr()

sns.heatmap(tips_corr)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Heatmap of Total Bill vs Tip")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
sns.boxplot(x="smoker", y="total_bill", data=tips)
plt.xlabel("Smoker")
```

```python
day_time_ratio = pd.crosstab(tips['sex'],tips['smoker'])

day_time_ratio.plot(kind='bar',stacked=False)

plt.title("Conditional Distribution of Smoker by Sex")
plt.xlabel("Sex")
plt.ylabel("Proportion")
plt.legend(title='Smoker')
plt.show()  
```
