```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips=sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
day_time_ratio = pd.crosstab(tips['day'],tips['time'])

day_time_ratio.plot(kind='bar',stacked=False)

plt.title("Conditional Distribution of Time by Day")
plt.xlabel("Day")
plt.ylabel("Proportion")
plt.legend(title='Time')
plt.show()  
```
