```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson,geom,hypergeom

np.random.seed(42)
```

```python
lam=4
n_samples=1000
poisson_samples=poisson.rvs(mu=lam,size=n_samples)
sample_mean = np.mean(poisson_samples)
sample_var = np.var(poisson_samples)


theoretical_mean = lam
theoretical_var = lam


print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')

x = np.arange(0, 10)
plt.hist(poisson_samples,bins=np.arange(0,10,1),density=True,alpha=0.6,edgecolor='black')
plt.plot(x,poisson.pmf(x,mu=lam),marker='o',linestyle='-',color='red',label='Theoretical PMF')
plt.title('Poisson Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()

```

```python
p=0.3
n_samples=1000
geom_samples=np.random.geometric(p,size=n_samples)

theoretical_mean = 1/p
theoretical_var = (1-p)/(p*p)
sample_mean = np.mean(geom_samples)
sample_var = np.var(geom_samples)

print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')

x = np.arange(0, 10)
plt.hist(geom_samples,bins=np.arange(0,10,1),density=True,alpha=0.6,edgecolor='black')
plt.plot(x,geom.pmf(x,p),marker='o',linestyle='-',color='red',label='Theoretical PMF')
plt.title('Geometric Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()
```

```python
N=50
n=20
k=10
n_samples=1000
hypergeom_samples=np.random.hypergeometric(ngood=n,nbad=N-n,nsample=k,size=n_samples)

theoretical_mean = n*k/N
theoretical_var = n*k/N*(1-(k/N))*((N-n)/(N-1))
sample_mean = np.mean(hypergeom_samples)
sample_var = np.var(hypergeom_samples)

print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 평균: {theoretical_mean:.4f}')
print(f'이론적 분산: {theoretical_var:.4f}')

x = np.arange(0, 10)
plt.hist(hypergeom_samples,bins=np.arange(0,10,1),density=True,alpha=0.6,edgecolor='black')
plt.plot(x,hypergeom.pmf(x,n,N,k),marker='o',linestyle='-',color='red',label='Theoretical PMF')
plt.title('Hypergeometric Distribution')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.show()
```

```python
from scipy.stats import binom,poisson
n=100
p=0.03
lam = n*p

x=np.arange(0,15)

pmf_binom = binom.pmf(x,n,p)
pmf_pois=poisson.pmf(x,lam)

plt.figure(figsize=(8,4))
plt.plot(x,pmf_binom,marker='o',linestyle='-',color='blue',label='Binomial PMF')
plt.plot(x,pmf_pois,marker='s',linestyle='--',color='red',label='Poisson PMF')
plt.xlabel('X')
plt.ylabel('Probability')
plt.title('Binomial and Poisson Distributions')
plt.legend()
plt.show()
```
